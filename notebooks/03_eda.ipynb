```python
import sys
from pathlib import Path

repo_root = Path.cwd().resolve().parent
sys.path.insert(0, str(repo_root / "src"))
```

```python
from rivwq.eda.io import load_clean_data

df = load_clean_data("../data/processed/dataset/data.csv")
core = ["Discharge", "Height", "DO", "fChl", "Temperature", "Turbidity"]
spectral = [
    "B1","B2","B3","B4","B5","B6","B7","B8","B8A","B9","B11","B12",
    "TCI_B","TCI_G","TCI_R","AOT","WVP",
    "MNDWI","GNDVI","SDDI","NDTI","BR","NDWI","NDPI","NDCI","2BDA_Chl","RR",
]
core = [c for c in core if c in df.columns]
spectral = [c for c in spectral if c in df.columns]
df.head().T
```

```python
from rivwq.eda.stats import descriptive_table, shapiro_table

desc = descriptive_table(df, cols=core + spectral)
shapiro_core = shapiro_table(df, cols=core)
shapiro_spectral = shapiro_table(df, cols=spectral)

desc.head()
```

```python
from rivwq.eda.stats import kendall_tau_matrix
from rivwq.eda.plots import set_style, corr_heatmap, distplots, qqplots
from rivwq.eda.stats import pearson_corr

set_style()

corr_kendall = kendall_tau_matrix(df, cols=core + spectral)

corr_heatmap(
    corr_kendall,
    out_path="../figures/eda/kendall_tau_heatmap.png",
    annot=False,
    mask_upper=True,
)

corr_pearson = pearson_corr(df, cols=core + spectral)

corr_heatmap(
    corr_pearson,
    out_path="../figures/eda/pearson_heatmap.png",
    annot=False,
    mask_upper=True,
)

distplots(df, core, "../figures/eda/dist_core.png", bins=30, kde=True)
distplots(df, spectral, "../figures/eda/dist_spectral.png", bins=30, kde=True)
qqplots(df, core, "../figures/eda/qq_core")
qqplots(df, spectral, "../figures/eda/qq_core")
```

```python
from rivwq.eda.plots import set_style, pairplot_kde_diag

pairplot_kde_diag(
    df, core,
    out_path="../figures/eda/pairplot_core.png",
    lower_kind="scatter",
    sample=None,
)
```

```python
import os
os.makedirs("../results/eda", exist_ok=True)

desc.to_csv("../results/eda/descriptive.csv")
shapiro_core.to_csv("../results/eda/shapiro_core.csv", index=False)
shapiro_spectral.to_csv("../results/eda/shapiro_spectral.csv", index=False)
corr_kendall.to_csv("../results/eda/kendall_tau_matrix.csv")
corr_pearson.to_csv("../results/eda/pearson_matrix.csv")

print("Exported EDA outputs to results/eda and figures/eda")
```
